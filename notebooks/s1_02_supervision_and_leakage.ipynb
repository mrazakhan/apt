### Why this problem matters

Loss masking and future-information leakage are distinct but equally important.

### Concepts and definitions

SFT prompt/completion examples need label masks that exclude unintended tokens. A prompt must not contain hidden verifier outputs or decisions from the future.

**Core mathematical object**

$$\mathrm{supervised\_fraction}=\frac{\#\{t:y_t\neq-100\}}{\#\{\text{sequence positions}\}}$$

Interpret every quantity with its denominator, conditioning context, measurement source and train/eval boundary; a formula is not evidence until those details are fixed.

### Worked example (separate from your exercise)

For a 100-token packed sequence, 60 are prompt, 30 target completion, and 10 padding. If exactly 30 tokens are supervised, mask fraction is 70% (under this stated convention).
